```python
import unicodedata
from nltk.tokenize import RegexpTokenizer
tokenizador = RegexpTokenizer(r"\w+")
def normalizar(texto):
  texto = texto.lower()
  texto = unicodedata.normalize("NFD", texto)
  texto = "".join(
    letra for letra in texto
    if unicodedata.category(letra) != "Mn"
  )
  palabras = tokenizador.tokenize(texto)
  return " ".join(palabras)
print(normalizar("¿DÓNDE está el MENÚ?"))
```

```python
intenciones = {
  "saludo": {
    "reglas": ["hola", "saludos", "buenos dias", "buenas tardes"],
    "respuesta": "Hola, podemos ayudarte con horarios, contactos, pedidos, catalogo y ubicaciones."
  },
  "horario": {
    "reglas": ["horario", "servicio", "a que hora abren"],
    "respuesta": "Nuestro horario de servicio es de 8am a 6pm de lunes a viernes."
  },
  "limite": {
    "reglas": ["limite para recibir pedidos"],
    "respuesta": "Nuestro horario de corte de pedidos es a las 4pm. Si estas fuera de este rango, te sugerimos contactes a tu ejecutivo"
  },
  "orden": {
    "reglas": ["quien puede tomar mi orden", "a quien le mando mi orden", "quien procesa las ordenes"],
    "respuesta": "Puedes hacer un pedido directamente llamanos al 81 8030 XXX o envia un correo a pedidos@xxx.com."
  },
  "credito": {
    "reglas": ["cuanto debo a la fecha", "cual es mi saldo", "cual es mi estado de cuenta"],
    "respuesta": "Puedes validar cual es tu estado de cuenta contacta directamente al departamento de Credito y Cobranza al 81 8030 XXXX o a cryco@xxx.com."
  },
  "pagos": {
    "reglas": ["cuales son los datos para pagar", "a donde mando mi transferencia"],
    "respuesta": "Podemos enviarte los datos particulares para hacer un pago contacta directamente al departamento de Crédito y Cobranza al 81 8030 XXXX o a cryco@xxx.com."
  },
  "facturacion": {
    "reglas": ["donde esta mi factura", "me mandan mi factura", "cual fue mi ultima factura"],
    "respuesta": "Puedes pedir una copia de tu factura, busca directamente al departamento de Facturacion al 81 8030 XXXX o a facturacion@xxx.com."
  }
}
print("Intenciones cargadas:", len(intenciones))
```

```python
intenciones.update({
  "correcion": {
    "reglas": ["mi factura esta incorrecta", "quiero corregir mi factura"],
    "respuesta": "Puedes pedir una copia de tu factura, busca directamente al departamento de Facturación al 81 8030 XXXX o a facturacion@xxx.com."
  },
  "pedido": {
    "reglas": ["donde esta mi pedido", "a que hora llega mi pedido", "cuando llega mi pedido"],
    "respuesta": "Puedes hacer tu solicitud directamente al equipo de Operaciones al 81 8030 XXXX o a operaciones@xxx.com."
  },
  "conmutador": {
    "reglas": ["cual es su telefono", "cual es el numero de su conmutador", "a quien le puedo llamar"],
    "respuesta": "Puedes llamarnos al 81 8030 XXXX."
  },
  "catalogo": {
    "reglas": ["a quien le puedo pedir un catalogo", "cual es su catalogo"],
    "respuesta": "Puedes pedirle tu catalogo directamente al equipo de Mercadotecnia al 81 8030 XXXX o a mercadotecnia@xxx.com."
  },
  "productos": {
    "reglas": ["que productos venden", "productos"],
    "respuesta": "Puedes pedirle tu catalogo directamente al equipo de Mercadotecnia al 81 8030 XXXX o a mercadotecnia@xxx.com."
  },
  "ubicacion": {
    "reglas": ["En donde estan ubicados", "cual es su ubicacion"],
    "respuesta": "Estamos ubicados en el Parque Industrial Landus en Santa Catarina, NL.."
  },
  "despedida": {
    "reglas": ["adios", "salir", "gracias", "bye"],
    "respuesta": "Gracias por contactarnos, que tengas un buen día. Esta sesión se está finalizando."
  },
})
print("Intenciones cargadas:", len(intenciones))
```

```python
def responder(mensaje):
  texto = normalizar(mensaje)
  if texto == "salir":
    return "salir", "Sesion finalizada."

  coincidencias = []
  for nombre, datos in intenciones.items():
    for regla in datos["reglas"]:
      patron = " " + normalizar(regla) + " "
      if patron in " " + texto + " ":
        coincidencias.append(nombre)
        break

  if len(coincidencias) == 1:
    nombre = coincidencias[0]
    return nombre, intenciones[nombre]["respuesta"]
  if len(coincidencias) > 1:
    return "ambigua", "Consulta un solo tema por mensaje."
  return "desconocida", "No entendi. Prueba: horario o menu."

print(responder("¿Cuál es el HORARIO?"))
```

```python
print("Proyecto MBIA_Fase1_Ontiveros. Escribe salir para cerrar.")
while True:
    mensaje = input("Tu: ")
    intencion, respuesta = responder(mensaje)
    print("Intencion:", intencion)
    print("Bot:", respuesta)
    if intencion == "salir":
      break
```
